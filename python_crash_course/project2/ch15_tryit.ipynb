15-1. Cubes: A number raised to the third power is a cube Plot the first five cubic numbers, and then plot the first 5000 cubic numbers

```python
#importing libraries
import matplotlib
import matplotlib.pyplot as plt

#plotting the first five cubic numbers
x_values = [1, 2, 3, 4, 5]
y_values = [x**3 for x in x_values]

plt.scatter(x_values, y_values, c = "green", edgecolor = "none", s = 39)
plt.show()
```

```python
#plotting the first 5000 cubic numbers
x_values = list(range(1, 1001))
y_values = [x**3 for x in x_values]

plt.scatter(x_values, y_values, c = "purple", edgecolor = "none", s = 22)
plt.show()
```

5-2. Colored Cubes: Apply a colormap to your cubes plot

```python
x_values = list(range(1, 1001))
y_values = [x**3 for x in x_values]

plt.scatter(x_values, y_values, c = y_values, cmap= plt.cm.managua, edgecolor = "none", s = 12)
plt.show()
```
